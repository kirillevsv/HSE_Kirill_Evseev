```python
import random
import time

# 1. Генерация массива с использованием range (случайный шаг от 3 до 5)
start_value = 10
end_value = 250_000_000
random_step = random.randint(3, 5)

print(f"Генерируем массив от {start_value} до {end_value} с шагом {random_step}...")
sorted_array = list(range(start_value, end_value, random_step))
print(f"Размер массива: {len(sorted_array):,} элементов")
print(f"Первый элемент: {sorted_array[0]}, последний: {sorted_array[-1]}")
print()

# 2. Генерация 10 случайных чисел с помощью list comprehensions и random.randint
# Генерируем числа в диапазоне массива, чтобы поиск был осмысленным
random_numbers = [random.randint(start_value, end_value) for _ in range(10)]
print(f"Сгенерированные случайные числа: {random_numbers}")
print()


# 3. Функция линейного поиска
def linear_search(arr, target):
    """
    Линейный поиск элемента в массиве.
    Возвращает индекс элемента или -1, если элемент не найден.
    """
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1


# 4. Функция бинарного поиска
def binary_search(arr, target):
    """
    Бинарный поиск элемента в отсортированном массиве.
    Возвращает индекс элемента или -1, если элемент не найден.
    """
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1


# 5. Проверка наличия случайных чисел в массиве
print("=" * 80)
print("РЕЗУЛЬТАТЫ ПОИСКА")
print("=" * 80)

# Замер времени для линейного поиска
print("\n--- ЛИНЕЙНЫЙ ПОИСК ---")
linear_start = time.time()
linear_results = []
for num in random_numbers:
    index = linear_search(sorted_array, num)
    linear_results.append(index)
    if index != -1:
        print(f"Число {num} найдено по индексу {index:,}")
    else:
        print(f"Число {num} НЕ найдено")
linear_end = time.time()

print(f"\nОбщее время линейного поиска: {linear_end - linear_start:.6f} секунд")
print(f"Среднее время на один поиск: {(linear_end - linear_start) / len(random_numbers):.6f} секунд")

# Замер времени для бинарного поиска
print("\n--- БИНАРНЫЙ ПОИСК ---")
binary_start = time.time()
binary_results = []
for num in random_numbers:
    index = binary_search(sorted_array, num)
    binary_results.append(index)
    if index != -1:
        print(f"Число {num} найдено по индексу {index:,}")
    else:
        print(f"Число {num} НЕ найдено")
binary_end = time.time()

print(f"\nОбщее время бинарного поиска: {binary_end - binary_start:.6f} секунд")
print(f"Среднее время на один поиск: {(binary_end - binary_start) / len(random_numbers):.6f} секунд")

# Проверка корректности результатов
print("\n--- ПРОВЕРКА КОРРЕКТНОСТИ ---")
all_correct = True
for i, num in enumerate(random_numbers):
    if linear_results[i] != binary_results[i]:
        print(f"⚠️ Расхождение для числа {num}: линейный={linear_results[i]}, бинарный={binary_results[i]}")
        all_correct = False

if all_correct:
    print("✅ Все результаты поиска совпадают!")

print(f"\n--- СРАВНЕНИЕ ПРОИЗВОДИТЕЛЬНОСТИ ---")
linear_time = linear_end - linear_start
binary_time = binary_end - binary_start
speedup = linear_time / binary_time if binary_time > 0 else float('inf')
print(f"Бинарный поиск быстрее линейного в {speedup:.1f} раз")
```
