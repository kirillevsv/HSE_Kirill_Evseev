```python
from datetime import datetime
from typing import Optional, List, Union
class CourtCase:
    def __init__(self, case_number: str):
        """
        Инициализация судебного дела.

        Args:
            case_number (str): Номер дела (обязательный параметр)
        """
        self.case_number = case_number
        self.case_participants: List[str] = []
        self.listening_datetimes: List[dict] = []
        self.is_finished: bool = False
        self.verdict: str = ""

    def set_a_listening_datetime(
        self,
        date: Union[str, datetime],
        time: Optional[str] = None,
        status: str = "scheduled",
        **kwargs
    ) -> None:
        """
        Добавляет судебное заседание в список listening_datetimes.

        Args:
            date (Union[str, datetime]): Дата заседания (строка в формате ГГГГ-ММ-ДД или объект datetime)
            time (Optional[str]): Время заседания в формате ЧЧ:ММ (опционально)
            status (str): Статус заседания (scheduled, in_progress, completed, cancelled)
            **kwargs: Дополнительные параметры (например, room, judge)
        """
        # Если date - строка, пробуем преобразовать в datetime
        if isinstance(date, str):
            try:
                date_obj = datetime.strptime(date, "%Y-%m-%d")
            except ValueError:
                # Если не удалось распарсить, сохраняем как есть
                date_obj = date
        else:
            date_obj = date

        # Создаем запись о заседании
        hearing = {
            "date": date_obj,
            "time": time,
            "status": status,
            **kwargs
        }

        self.listening_datetimes.append(hearing)

    def add_participant(self, inn: str) -> None:
        """
        Добавляет участника в список case_participants.

        Args:
            inn (str): ИНН участника
        """
        # Проверяем, что ИНН не пустой и не None
        if not inn:
            raise ValueError("ИНН участника не может быть пустым")

        # Проверяем, что участник еще не добавлен
        if inn not in self.case_participants:
            self.case_participants.append(inn)

    def remove_participant(self, inn: str) -> bool:
        """
        Убирает участника из списка case_participants.

        Args:
            inn (str): ИНН участника для удаления

        Returns:
            bool: True если участник был удален, False если не найден
        """
        if inn in self.case_participants:
            self.case_participants.remove(inn)
            return True
        return False

    def make_a_decision(self, verdict: str) -> None:
        """
        Выносит решение по делу.

        Args:
            verdict (str): Текст решения по делу
        """
        if self.is_finished:
            raise ValueError("Дело уже завершено, решение не может быть изменено")

        if not verdict or not verdict.strip():
            raise ValueError("Решение не может быть пустым")

        self.verdict = verdict
        self.is_finished = True

    def __str__(self) -> str:
        """Строковое представление дела."""
        status = "Завершено" if self.is_finished else "В процессе"
        participants_count = len(self.case_participants)
        hearings_count = len(self.listening_datetimes)

        return (f"Дело №{self.case_number} | Статус: {status} | "
                f"Участников: {participants_count} | Заседаний: {hearings_count}")

    def __repr__(self) -> str:
        """Представление для разработчиков."""
        return f"CourtCase(case_number='{self.case_number}')"
```
